```python
%matplotlib notebook

import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0, 1, 15)
r1 = 10
r2 = 50

x = ((r2 - r1)* t + r1) * np.cos(t*4*np.pi)
y = ((r2 - r1)* t + r1) * np.sin(t*4*np.pi)
z = 0*t
```

```python
f1 = plt.figure()
f1.add_subplot(projection="3d")
plt.plot(x, y, z);
```

```python
for i in range(1, len(t)):
    d = np.sqrt((x[i] - x[i-1])**2 + (y[i] - y[i-1])**2)
    print(d/0.34)
```

```python
%matplotlib notebook

import numpy as np
import matplotlib.pyplot as plt
```

```python
t = np.linspace(0, 1, 15)
r1 = 10
r2 = 30

x = ((r2 - r1)* t + r1) * np.cos(t*4*np.pi)
y = ((r2 - r1)* t + r1) * np.sin(t*4*np.pi)
z = np.linspace(-50, 0, 15)
```

```python
f1 = plt.figure()
f1.add_subplot(projection="3d")
plt.plot(x, y, z);
```

```python
for i in range(1, len(t)):
    d = np.sqrt((x[i] - x[i-1])**2 + (y[i] - y[i-1])**2 + (z[i] - z[i-1])**2)
    print(d/0.34)
```

```python
with open("rcoil_points.txt", 'w') as f:
    for i in range(len(x)):
        f.write("{:.2f}, {:.2f}, {:.2f}\n".format(x[i], y[i], z[i]))
```

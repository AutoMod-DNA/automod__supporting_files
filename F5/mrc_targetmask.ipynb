```python
import numpy as np
import mrcfile
```

```python
f_path = 'cryosparc_P3_J72_map_sharp.mrc'

with mrcfile.open(f_path) as mrc:
    print(mrc.data.shape)
    print(mrc.voxel_size)
    print(mrc.data.dtype)
    print(mrc.header)
    old_header = mrc.header
    old_vsize = mrc.voxel_size
```

```python
mask_tight = np.zeros((256, 256, 256), dtype=np.float32)
# interhelical distance 2.5 nm
mask_loose = np.zeros((256, 256, 256), dtype=np.float32)
# interhelical distance 5.0 nm

for i in range(256):
    if i in [j for j in range(110, 146)]:
        for j in range(256):
            for k in range(256):
                d = np.sqrt(np.abs(j-127)**2 + np.abs(k-127)**2)
                if 45 <= d <= 62:
                    mask_tight[i, j, k] = 1
                    
for i in range(256):
    if i in [j for j in range(93, 163)]:
        for j in range(256):
            for k in range(256):
                d = np.sqrt(np.abs(j-127)**2 + np.abs(k-127)**2)
                if 39 <= d <= 69:
                    mask_loose[i, j, k] = 1
```

```python
with mrcfile.new('3Dtarget_tight_as_binary_mask.mrc', overwrite=True) as mrc:
    mrc.set_data(mask_tight)
    mrc.voxel_size = old_vsize
    mrc.set_extended_header(old_header)
    
with mrcfile.new('3Dtarget_loose_as_binary_mask.mrc', overwrite=True) as mrc:
    mrc.set_data(mask_loose)
    mrc.voxel_size = old_vsize
    mrc.set_extended_header(old_header)
```

```python
with mrcfile.open('3Dtarget_tight_as_binary_mask.mrc') as mrc:
    print(mrc.data.shape)
    print(mrc.voxel_size)
    print(mrc.data.dtype)
    print(mrc.header)
    
with mrcfile.open('3Dtarget_loose_as_binary_mask.mrc') as mrc:
    print(mrc.data.shape)
    print(mrc.voxel_size)
    print(mrc.data.dtype)
    print(mrc.header)
```
